```python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5)+1):
        if n % i == 0:
            return False
    return True
```

```python
def remove_duplicates(lst):
    return list(dict.fromkeys(lst))  # Сохраняет порядок
```

```python
def unique_chars(s):
    return len(set(s))
```

```python
def list_intersection(lst1, lst2):
    return list(set(lst1) & set(lst2))
```
