```python
def check_any_true(lst):
    return any(all(sublist) for sublist in lst)
```

```python
def validate_types(objects, types):
    return all(isinstance(obj, types) for obj in objects)
```

```python
def compare_objects(a, b):
    if a is b:
        return "Объекты идентичны"
    elif a == b:
        return "Содержимое одинаково, но объекты разные"
    return "Объекты разные"
```

```python
def get_object_info(obj):
    return [attr for attr in dir(obj) if not attr.startswith('__')]
```

```python
def check_all_true(lst):
    return all(any(sublist) for sublist in lst)
```

```python
def power(base, exponent):
    if exponent == 0:
        return 1
    return base * power(base, exponent - 1)
```

```python
def count_down(n):
    if n < 0:
        return
    print(n, end=', ' if n > 0 else '')
    count_down(n - 1)
```

```python
def find_max(lst):
    if len(lst) == 1:
        return lst[0]
    return max(lst[0], find_max(lst[1:]))
```

```python
def reverse_string(s):
    if len(s) == 0:
        return s
    return reverse_string(s[1:]) + s[0]
```

```python
def flatten_list(lst):
    result = []
    for item in lst:
        if isinstance(item, list):
            result.extend(flatten_list(item))
        else:
            result.append(item)
    return result
```

```python
def outer():
    x = 10
    def inner():
        nonlocal x
        x = 20
    inner()
    return x
```

```python
counter = 0

def global_counter():
    global counter
    counter += 1
```
